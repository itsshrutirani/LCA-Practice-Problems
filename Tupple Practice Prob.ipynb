```python
#4.FIND INDEX OF A ITEM IN A LIST
essentials=('charger','tissue box','umbrella','earphones','water bottle')
item=input("Enter item you want to find:").lower()
print(item)
for i in range(len(essentials)):
    if essentials[i].lower()==item:
        print("Index:",essentials.index(item))
        break
else:    
    print("Item not found!")
```

```python
#5.REPLACE LAST VALUE OF TUPLE
num_list=[(10,20,30),(40,50,60),(70,80,90)]

for i in range(len(num_list)):
    num_list[i]=num_list[i][:-1]+(100,)
print(num_list)
```
