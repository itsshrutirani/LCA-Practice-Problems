```python
# 4.CHECK POSITIVE,NEGATIVE OR ZERO
def check_num(n):
    if n>0:
        return"Given number is positive!"
    elif n==0:
        return"Given number is zero!"
    else:
        return"Given number is Negative!"

n1=6
print(check_num(n1))

n2=0
print(check_num(n2))

n3=-8
print(check_num(n3))
```

```python
# 5.CHECK IF EVEN OR ODD
def check_even(num):
    if num%2==0:
        return "Given number is EVEN!"
    else:
        return "Given number is ODD!"

num1=66
print(check_even(num1))

num2=45
print(check_even(num2))
```

```python
# 9.CHECK IF GIVEN NUMBER IS PRIME OR NOT
def check_prime(num):
    for i in range(2,num):
        if num%i==0:
            return"Number is NOT PRIME!"
    return"Number is PRIME!"

num1=35
print(check_prime(num1))

num2=7
print(check_prime(num2))
```

```python
# 10.PRINT PRIME NUMBERS BETWEEN 10 AND 99
for num in range(10,100):
    prime=True

    for i in range(2,num):
        if num%i==0:
            prime=False
            break
    if prime:
        print(num)
```
