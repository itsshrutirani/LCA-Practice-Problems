```python
#1.ADD A KEY VALUE PAIR
my_dict={
    0:10,
    1:20
}
my_dict[2]=30
print(my_dict)
```

```python
#2.CONCATENATE DICTIONARIES
dict1={
    1:10,
    2:20}

dict2={
    3:30,
    4:40}

dict3={
    5:50,
    6:60}

new_dict={}

new_dict.update(dict1)
new_dict.update(dict2)
new_dict.update(dict3)
print(new_dict)
```

```python
#3.CHECK IF KEY ALREADY EXISTS
my_dict = {1: 10, 2: 20, 3: 30}
key=int(input("Enter key:"))
print(key)
if key in my_dict:
    print("Key exists")
else:
    print("Key does not exist")
    
```

```python
#4.PRINT KEYS,VALUES AND KEY-VALUE PAIRS
my_dict={
    "Name":"Shruti Rani",
    "Age":19,
    "course":"CSE(AI & DS)",
    "Div":9
}

print("\nKEYS:")
for key in my_dict:
    print(key)

print("\nVALUES:")
for values in my_dict:
    print(values)

print("\nKEYS-VALUES PAIRS:")
for items in my_dict:
    print(items)
```

```python
#5.PREPARE A DICTIONARY WITH NUMBERS 1-15 AND THIER SQUARES
num_dict={}
for i in range(1,16):
    num_dict[i]=i**2
print(num_dict)

```

```python
#6.SUM OD DICTIONARY VALUES
dictionary={
    'a':10,
    'b':20,
    'c':30,
    'd':40
}
values=dictionary.values()
sum=0
for val in values:
    sum+=val

print(sum)
```

```python
#9.DICTIONARY TO LISTS
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34
}

keys = list(my_dict.keys())
values = list(my_dict.values())

print("Keys:", keys)
print("Values:", values)
```

```python
#10.MEAN VALUES OF DICTIONARY
my_dict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

total = 0

for value in my_dict.values():
    total = total + value

mean = total / len(my_dict)

print("Mean:", mean)
```
