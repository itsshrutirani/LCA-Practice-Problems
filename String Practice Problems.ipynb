```python
#1.COUNT NO OF UPPER CASE AND LOWER CASE LETTER
text=input("Enter your text here:")

upper=0
lower=0

for character in text:
    if character.islower():
        lower+=1
    elif character.isupper():
        upper+=1
print("Lower Case letters:",lower)
print("Upper Case letters:",upper)
```

```python
#2.CHECK IF A STRING IS PALINDROME OR NOT
word=input("Enter your word:")
letters=word.split()
def check_palindrome(letters):
    letter_list=list(letters)
    letter_list.reverse()
    if letter_list==letters:
        return"It is a Palindrome!"
    else:
        return"It is not a Palindrome!"

print(word)
print(check_palindrome(letters))
```

```python
#3.REPEAT FIRST TWO CHARACTERS N TIMES
text=input("Enter a string:")
n=len(text)
word=text[:2]*n
print(word)
```

```python
#4.REMOVE X IF IT OCCURS AT THE BEGGINING OR END
text=input("Enter a string:")
if text[0]=="x":
    text=text[1:]
if text[-1]=="x":
    text=text[:-1]
print(text)
    
```

```python
#5.REPEAT THE LAST N CHARACTERS N TIMES
text = input("Enter a string: ")

n = int(input("Enter number:"))

result = text[-n:] * n

print(result)
```
