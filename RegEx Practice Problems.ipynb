```python
#1.RECOGNIZE CERTAIN WORDS
import re

text = input("Enter a word: ")

pattern = r"^[bh][aiu]t$"

if re.match(pattern, text):
    print("Match found")
else:
    print("No match")
```

```python
#2.MATCH TWO WORDS SEPARATED BY EXACTLY ONE SPACE
import re

text = input("Enter two words: ")

pattern = r"^[A-Za-z]+ [A-Za-z]+$"

if re.match(pattern, text):
    print("Match found")
else:
    print("No match")
```

```python
#3.MATCH A WORD AND A SINGLE LETTER SEPARATED BY COMMA AND A SPACE
import re

text = input("Enter word and letter: ")

pattern = r"^[A-Za-z]+, [A-Za-z]$"

if re.match(pattern, text):
    print("Match found")
else:
    print("No match")
```
