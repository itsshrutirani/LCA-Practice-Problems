```python
#1. REMOVE A ITEM FROM SET
my_set = {10, 20, 30, 40, 50}

num = int(input("Enter item to remove: "))

my_set.remove(num)

print(my_set)
```

```python
#2.INTERSECTION OF TWO SETS
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

result = set1.intersection(set2)

print(result)
```

```python
#3.UNION OF TWO SETS
set1 = {10, 20, 30}
set2 = {30, 40, 50}

result = set1.union(set2)

print(result)
```

```python
#4.FIND MAX AND MIN IN SET
my_set = {10, 25, 5, 40, 15}

print("Maximum:", max(my_set))
print("Minimum:", min(my_set))
```
