```python
# 9.ACCEPT VALUE AND PERFORM OPERATIONS
my_list=[]
for i in range(20):
    nums=int(input("Enter your number:"))
    my_list.append(nums)

print(my_list)
# a) count similar elements of list and print their index value
for num in set(my_list):
    indexes=[]

    for i in range(len (my_list)):
        if my_list[i]==num:
            indexes.append(i)

    if len(indexes)>1:
        print(num,"occurs",len(indexes),"times at indexes:",indexes)

#b) count even and odd values of list
even=0
odd=0
for num in my_list:
    if num%2==0:
        even+=1
    else:
        odd+=1

print("Even Numbers:",even)
print("Odd Numbers:",odd)

#c) count positive and negative values of list.
positives=0
negatives=0

for num in my_list:
    if num>=0:
        positives+=1
    else:
        negatives+=1

print("Positives:",positives)
print("Negatives:",negatives)
```

```python
# 10.LIST OPERATIONS
my_list=[]
for i in range(10):
    nums=int(input("Enter Numbers:"))
    my_list.append(nums)
print("Original List:",my_list)

# a)Sort list in ascending order using sorted() function and display sorted list
my_list.sort()
print("\nList in asscending order:",my_list)

#b)sort list in descending order using sort() function
my_list.sort(reverse=True)
print("\nList in decending order:",my_list)

#c)display length of list
print("\nlength of list is:",len(my_list))
```

```python
# 11.ACCEPT TWO LISTS AND MERGE THEM
list1 = []
list2 = []

for i in range(5):
    num = int(input("Enter value for list 1: "))
    list1.append(num)

for i in range(5):
    num = int(input("Enter value for list 2: "))
    list2.append(num)

list3 = list1 + list2
print("list1:",list1)
print("list2:",list2)
print("list3:",list3)
```

```python
#12.ACRONYM
phrase=input("Enter your phrase: ")
words=phrase.split()
acronym=''
for word in words:
    acronym+=word[0]
print("phrase:",phrase)
print(acronym.upper())
```
